```python
import pandas as pd
import numpy as np
import os
import seaborn as sns
import plotly.express as px
import matplotlib as plt

os.chdir('/users/maria/downloads')

df = pd.read_csv('ventas_catering.csv')
```

```python
df = df.drop(['Unnamed: 11','65'], axis=1)
```

```python
df = df.fillna('-')
```

```python
df['Dia'] = pd.to_datetime(df['Dia'], dayfirst=True)
```

```python
df['Total'] = df['Total'].astype(str).str.replace('$', '').str.replace(',', '').str.strip()
df['Total'] = pd.to_numeric(df['Total'], errors='coerce')
```

```python
# Extraer el número del mes (1 al 12)
df['Mes'] = df['Dia'].dt.month

# Extraer el número de la semana del año
df['Semana'] = df['Dia'].dt.isocalendar().week
```

## Visualización de datos para análisis

```python
df_ventas_producto = df.groupby(['Producto','Semana'])['Cantidad'].sum().reset_index()
```

```python
fig = px.histogram(df_ventas_producto, x= 'Semana', y='Cantidad',title='Cantidad de productos por semana',color='Producto', text_auto='.2s', nbins=20)
fig.show()
```

```python
df_ultimas_semanas = df_ventas_producto[df_ventas_producto['Semana'] >= 16]
```

```python
fig = px.histogram(df_ultimas_semanas, x= df_ultimas_semanas['Semana'].astype(str), y='Cantidad',title='Cantidad de productos ultimas 6 semanas',color='Producto', text_auto='.2s', nbins=20)
fig.show()
```

## Gráfico de Correlación

```python
columnas_corr = ['Cantidad', 'Total', 'Precio de venta']

df_corr = df[columnas_corr]
```

```python
fig = px.imshow(df_corr.corr(numeric_only=True), text_auto=True, title='Lolo Munch Ventas - Correlación entre Cantidad, Total y Precio de venta')
fig.show()
```
